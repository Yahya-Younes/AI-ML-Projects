```python
import pandas as pd
import numpy as np
```

```python
teams = pd.read_csv("data/teams.csv")
```

```python
# Visualize the data 
teams
```

```python
X = teams[["athletes","prev_medals"]].copy()
y = teams[["medals"]].copy()
```

```python
X ["intercepts"] = 1
```

```python
X = X[["intercepts", "athletes","prev_medals"]]
```

```python
X_Transpose = X.T
```

```python
B = np.linalg.inv(X_Transpose @ X) @ X_Transpose @ y
```

```python
B.index = X.columns
```

```python
B
```

```python
predictions = X @ B
```

```python
predictions
```

```python
# Lets calculate R_squared = 1 - (SSR/SST) to see if there are other factors that we didn't take into consideration
```

```python
SSR = ((y - predictions) **2).sum()
```

```python
SST = ((y - y.mean()) **2).sum()
```

```python
SSR
```

```python
SST
```

```python
r_squared = 1 - (SSR/SST)
```

```python
r_squared
```
